## Задача 821. Shortest Distance to a Character ([ссылка](https://leetcode.com/problems/shortest-distance-to-a-character/description/)) [1 балл]

Перевод задачи:

Дана строка `s` и символ `c`, который встречается в `s`.

Нужно вернуть массив целых чисел `answer` длины `s.length`, где для каждого индекса `i` значение `answer[i]` равно расстоянию от индекса `i` до ближайшего вхождения символа `c` в строке `s`.  
Расстояние между двумя индексами `i` и `j` определяется как `|i - j|`, где `|x|` обозначает абсолютное значение `x`.

---

## Примеры

### Пример 1:
**Вход:**
```python
s = "loveleetcode"
c = "e"
```
**Выход:**
```
[3,2,1,0,1,0,0,1,2,2,1,0]
```
**Объяснение:**
Символ `'e'` встречается по индексам 3, 5, 6 и 11 (индексация с нуля).  
- Для индекса 0 ближайшее вхождение `'e'` находится по индексу 3, расстояние = `|0 - 3| = 3`.  
- Для индекса 1 ближайшее вхождение `'e'` находится по индексу 3, расстояние = `|1 - 3| = 2`.  
- Для индекса 4 имеется равное расстояние до вхождений `'e'` на индексах 3 и 5, расстояние = `1`.  
- Для индекса 8 ближайшее вхождение `'e'` находится по индексу 6, расстояние = `|8 - 6| = 2`.

### Пример 2:
**Вход:**
```python
s = "aaab"
c = "b"
```
**Выход:**
```
[3,2,1,0]
```

---

## Ограничения:
- $1 \leq s.length \leq 10^4$
- Символы строки `s` и символ `c` являются строчными латинскими буквами.
- Гарантируется, что символ `c` встречается хотя бы один раз в строке `s`.

````python
# Асимптотическая сложность по времени - О(n)
# Асимптотическая сложность по памяти - O(n)

# ВАШ КОД

class Solution:
    def shortestToChar(self, s: str, c: str) -> List[int]:
        n = len(s)
        answer = [0] * n
        prev = float('-inf')

        # Первый проход слева направо
        for i in range(n):
            if s[i] == c:
                prev = i
            answer[i] = i - prev if prev != float('-inf') else float('inf')

        # Второй проход справа налево
        prev = float('inf')
        for i in range(n - 1, -1, -1):
            if s[i] == c:
                prev = i
            answer[i] = min(answer[i], prev - i)

        return answer
````

## Задача 844. Backspace String Compare ([ссылка](https://leetcode.com/problems/backspace-string-compare/)) [1 балл]

Перевод задачи:

Даны две строки `s` и `t`. Необходимо определить, равны ли они после симуляции ввода в текстовый редактор, где символ `#` означает удаление последнего введённого символа (backspace). Если при выполнении backspace текст остается пустым, он остаётся пустым.

---

## Примеры

### Пример 1:
**Вход:**
```python
s = "ab#c"
t = "ad#c"
```
**Выход:**
```
true
```
**Объяснение:**
После обработки обе строки становятся `"ac"`.

### Пример 2:
**Вход:**
```python
s = "ab##"
t = "c#d#"
```
**Выход:**
```
true
```
**Объяснение:**
После обработки обе строки становятся пустыми.

### Пример 3:
**Вход:**
```python
s = "a#c"
t = "b"
```
**Выход:**
```
false
```
**Объяснение:**
После обработки строка `s` становится `"c"`, а строка `t` — `"b"`.

---

## Ограничения:
- $1 \leq s.length, t.length \leq 200$
- Строки `s` и `t` содержат только строчные латинские буквы и символы `#`.

````python
# Асимптотическая сложность по времени - О(n)
# Асимптотическая сложность по памяти - O(n)

# ВАШ КОД

class Solution:
    def backspaceCompare(self, s: str, t: str) -> bool:
        def build(string):
            stack = []
            for ch in string:
                if ch == '#':
                    if stack:
                        stack.pop()
                else:
                    stack.append(ch)
            return stack

        return build(s) == build(t)

````

## Задача 80. Remove Duplicates from Sorted Array II ([ссылка](https://leetcode.com/problems/remove-duplicates-from-sorted-array-ii/description/)) [1 балл]


Перевод задачи:

Дан отсортированный массив целых чисел `nums`. Требуется удалить дубликаты из массива «на месте» таким образом, чтобы каждый уникальный элемент встречался не более двух раз, сохраняя относительный порядок элементов. Изменять длину массива нельзя, поэтому результат нужно поместить в первые `k` позиций массива `nums`. Формально, если после удаления дубликатов осталось `k` элементов, первые `k` элементов `nums` должны содержать окончательный результат. Остальные элементы массива могут иметь любое значение.

Необходимо вернуть значение `k` после размещения окончательного результата в первых `k` ячейках массива `nums`.

**Условия:**
- Нельзя использовать дополнительное пространство для создания нового массива (требуется модификация входного массива «на месте» с использованием O(1) дополнительной памяти).

**Custom Judge:**

Решение проверяется следующим кодом:

```java
int[] nums = [...]; // Входной массив
int[] expectedNums = [...]; // Ожидаемый результат с правильной длиной

int k = removeDuplicates(nums); // Вызов вашей реализации

assert k == expectedNums.length;
for (int i = 0; i < k; i++) {
    assert nums[i] == expectedNums[i];
}
```

Если все утверждения (assert) пройдут, ваше решение будет принято.

---

## Примеры

### Пример 1:
**Вход:**
```python
nums = [1,1,1,2,2,3]
```
**Выход:**
```
5, nums = [1,1,2,2,3,_]
```
**Объяснение:**
Функция должна вернуть `k = 5`, а первые пять элементов массива `nums` должны быть равны `1, 1, 2, 2, 3`. Остальные элементы массива могут быть любыми (обозначены как `_`).

### Пример 2:
**Вход:**
```python
nums = [0,0,1,1,1,1,2,3,3]
```
**Выход:**
```
7, nums = [0,0,1,1,2,3,3,_,_]
```
**Объяснение:**
Функция должна вернуть `k = 7`, а первые семь элементов массива `nums` должны быть равны `0, 0, 1, 1, 2, 3, 3`. Остальные элементы массива могут быть любыми (обозначены как `_`).

---

## Ограничения:
- $1 \leq nums.length \leq 3 \times 10^4$
- $-10^4 \leq nums[i] \leq 10^4$
- Массив `nums` отсортирован по неубыванию.

````python
# Асимптотическая сложность по времени - О(n)
# Асимптотическая сложность по памяти - O(1)

# ВАШ КОД
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        if len(nums) <= 2:
            return len(nums)

        insert_pos = 2  # Первые два элемента можно оставить

        for i in range(2, len(nums)):
            if nums[i] != nums[insert_pos - 2]:
                nums[insert_pos] = nums[i]
                insert_pos += 1

        return insert_pos
````

## Задача B. Общая длина ([ссылка](https://codeforces.com/edu/course/2/lesson/9/3/practice/contest/307094/problem/B)) [0.33 балла]


Дан массив из 𝑛 чисел 𝑎𝑖.

Отрезок этого массива 𝑎[𝑙..𝑟] (где 1 ≤ 𝑙 ≤ 𝑟 ≤ 𝑛) называется **хорошим**, если сумма чисел на этом отрезке не больше 𝑠.

**Задача:** Найти сумму длин всех хороших отрезков.

---

## Входные данные

- Первая строка содержит два целых числа 𝑛 и 𝑠 (1 ≤ 𝑛 ≤ 10⁵, 1 ≤ 𝑠 ≤ 10¹⁸).
- Вторая строка содержит 𝑛 целых чисел 𝑎𝑖 (1 ≤ 𝑎𝑖 ≤ 10⁹).

---

## Выходные данные

Выведите одно число — сумму длин всех хороших отрезков.

---

## Пример

**Входные данные:**
```
7 20
2 6 4 3 6 8 9
```

**Выходные данные:**
```
39
```

````python
# Асимптотическая сложность по времени - О(n)
# Асимптотическая сложность по памяти - O(1) (можно в теории работать с потоком данных, а не хранить все в памяти)

# ВАШ КОД

n, s = map(int, input().split())
a = list(map(int, input().split()))

total = 0
current_sum = 0
r = 0

for l in range(n):
    while r < n and current_sum + a[r] <= s:
        current_sum += a[r]
        r += 1
    # Сумма длин всех отрезков начиная с l до r-1
    length = r - l
    total += length * (length + 1) // 2
    current_sum -= a[l]

print(total)

````

## Задача C. Город Че ([ссылка](https://codeforces.com/edu/course/2/lesson/9/3/practice/contest/307094/problem/C)) [0.33 балла]

В городе Че есть пешеходная улица — одно из самых популярных мест для прогулок. Вдоль этой улицы расположено 𝑛 забавных памятников. Маша, девочка из города Че, не может сделать выбор между двумя мальчиками из своей школы и решила назначить им свидание одновременно. Она планирует выбрать два памятника, возле которых мальчики будут ждать её, при этом она хочет, чтобы мальчики не увидели друг друга. Маша знает, что из-за тумана мальчики увидят друг друга, только если расстояние между ними не превышает 𝑟 метров.

**Задача:** Определить, сколькими способами можно выбрать два различных памятника так, чтобы мальчики не увидели друг друга (то есть расстояние между выбранными памятниками было строго больше 𝑟).

---

## Входные данные

- Первая строка содержит два целых числа 𝑛 и 𝑟 (2 ≤ 𝑛 ≤ 300000, 1 ≤ 𝑟 ≤ 10⁹) — количество памятников и максимальное расстояние, на котором мальчики могут увидеть друг друга.
- Во второй строке заданы 𝑛 положительных чисел 𝑑₁, …, 𝑑ₙ, где 𝑑ᵢ — расстояние от i-го памятника до начала улицы. Все памятники находятся на разных расстояниях от начала улицы и перечислены в порядке возрастания (1 ≤ 𝑑₁ < 𝑑₂ < … < 𝑑ₙ ≤ 10⁹).

---

## Выходные данные

Выведите одно число — количество способов выбрать два памятника для организации свиданий.

---

## Пример

**Входные данные:**
```
4 4
1 3 5 8
```

**Выходные данные:**
```
2
```

**Примечание:**
В данном примере Маша может выбрать памятники с расстояниями:
- 1 и 8 (памятники 1 и 4)
- 3 и 8 (памятники 2 и 4)

````python
# Асимптотическая сложность по времени - О(n)
# Асимптотическая сложность по памяти - O(1), если читать данные с потока, то можно хранить только переменые

n, r = map(int, input().split())
d = list(map(int, input().split()))

total = 0
right = 0

for left in range(n - 1):
    right = max(right, left + 1)
    while right < n and d[right] - d[left] <= r:
        right += 1
    total += n - right  # все пары с right и правее валидны

print(total)


# ВАШ КОД
````

Известная задача, просто делаем два указателя от 0 до n - 1

## Задача A. Зацикленный плейлист ([ссылка](https://codeforces.com/edu/course/2/lesson/9/3/practice/contest/307094/problem/A)) [0.33 балла]

Миша слушает музыку на плеере. Его плейлист состоит из n песен, которые воспроизводятся в определённом порядке, причем после окончания последней песни начинает играть первая (циклическое воспроизведение). Каждая песня имеет свою характеристику — позитивность a_i (натуральное число). После прослушивания i-й песни настроение Миши увеличивается на a_i.

Миша может начать прослушивание с любой песни и прослушать подряд любое количество песен, при этом некоторые песни могут быть прослушаны более одного раза.

Миша чувствует себя счастливым, если суммарное увеличение его настроения после прослушивания выбранного отрезка песен составит не меньше p. Он хочет добиться этого, прослушав как можно меньше песен. Помогите ему определить, с какой песни следует начать прослушивание, и сколько песен нужно прослушать, чтобы достичь желаемого повышения настроения.

---

## Входные данные

- Первая строка содержит два целых числа n и p (1 ≤ n ≤ 1000, 1 ≤ p ≤ 10¹⁸) — число песен в плейлисте и требуемое увеличение настроения.
- Вторая строка содержит n целых неотрицательных чисел a_i (1 ≤ a_i ≤ 10⁹) — позитивности песен.

---

## Выходные данные

Выведите два числа:
- k — номер песни, с которой Мише следует начать прослушивание;
- c — количество песен, которые нужно прослушать для достижения повышения настроения не меньше p.

Если существует несколько решений, выведите любое.

---

## Примеры

### Пример 1:
**Вход:**
```
9 10
1 2 3 4 5 4 3 2 1
```

**Выход:**
```
3 3
```

### Пример 2:
**Вход:**
```
5 6
3 1 1 1 4
```

**Выход:**
```
5 2
```

### Пример 3:
**Вход:**
```
3 100
10 10 10
```

**Выход:**
```
1 10
```


````python
# Асимптотическая сложность по времени - О(n lon(n))
# Асимптотическая сложность по памяти - O(n)


import sys, bisect

def main():
    data = sys.stdin.read().split()
    if not data:
        return
    it = iter(data)
    n = int(next(it))
    p = int(next(it))
    a = [int(next(it)) for _ in range(n)]

    # Построим массив префиксных сумм для одного цикла
    # prefix[0] = 0, prefix[i] = a[0] + ... + a[i-1] для 1 <= i <= n
    prefix = [0] * (n + 1)
    for i in range(1, n + 1):
        prefix[i] = prefix[i-1] + a[i-1]
    T = prefix[n]  # сумма одного полного цикла

    best = 10**20  # большое число
    best_start = -1

    # Перебираем возможные старты i (0-indexed)
    for i in range(n):
        # Сумма песен от a[i] до конца цикла
        partial = T - prefix[i]
        if partial >= p:
            # Всё укладывается в один цикл.
            # Ищем минимальное j из диапазона [i+1, n] такое, что:
            # prefix[j] - prefix[i] >= p
            target = p + prefix[i]
            j = bisect.bisect_left(prefix, target, i+1, n+1)
            candidate = j - i
        else:
            # Нужно прослушать не только остаток цикла, но и дополнительные полные циклы
            remaining = p - partial
            # Пусть k = количество полных циклов, которое нужно добавить, и затем дополнительно r песен из следующего цикла.
            k = remaining // T
            rem = remaining % T
            if rem == 0:
                # Если оставшаяся сумма делится на T, то достаточно k полных циклов,
                # и дополнительная часть равна 0.
                candidate = (n - i) + k * n
            else:
                # Ищем минимальное r (от 0 до n), такое что prefix[r] >= rem
                r = bisect.bisect_left(prefix, rem)
                candidate = (n - i) + k * n + r
        if candidate < best:
            best = candidate
            best_start = i

    # Выводим 1-индексированный номер песни и число песен
    sys.stdout.write(f"{best_start+1} {best}")

if __name__ == '__main__':
    main()


# ВАШ КОД
````

## Задача E. Рюкзак на отрезке. ([ссылка](https://codeforces.com/edu/course/2/lesson/9/3/practice/contest/307094/problem/E)) [0.5 балла]

Дан массив из 𝑛 предметов, для каждого из которых известны его вес 𝑤𝑖 и стоимость 𝑐𝑖.

**Задача:** Выбрать отрезок (подмассив) этого массива, для которого выполняются следующие условия:
- Суммарный вес предметов на отрезке не превышает 𝑠.
- Суммарная стоимость предметов на отрезке максимально возможная среди всех отрезков с суммарным весом не больше 𝑠.

Необходимо вывести одно число — наибольшую суммарную стоимость предметов, которые можно поместить в рюкзак, выбирая некоторый отрезок.

---

## Входные данные

- Первая строка содержит два целых числа 𝑛 и 𝑠 (1 ≤ 𝑛 ≤ 10⁵, 1 ≤ 𝑠 ≤ 10⁹) — количество предметов и максимальный допустимый суммарный вес.
- Во второй строке содержится 𝑛 целых чисел 𝑤𝑖 (1 ≤ 𝑤𝑖 ≤ 10⁹) — веса предметов.
- В третьей строке содержится 𝑛 целых чисел 𝑐𝑖 (1 ≤ 𝑐𝑖 ≤ 10⁹) — стоимости предметов.

---

## Выходные данные

Выведите одно число — наибольшую стоимость предметов, которые можно получить, выбрав отрезок с суммарным весом не больше 𝑠.

---

## Пример

**Входные данные:**
```
6 20
9 7 6 5 8 4
7 1 3 6 8 3
```

**Выходные данные:**
```
17
```

````python
# Асимптотическая сложность по времени - О(n)
# Асимптотическая сложность по памяти - O(1) если читать данные в потоке

n, s = map(int, input().split())
w = list(map(int, input().split()))
c = list(map(int, input().split()))

max_value = 0
current_weight = 0
current_value = 0
l = 0

for r in range(n):
    current_weight += w[r]
    current_value += c[r]

    # Если суммарный вес > s, сдвигаем левый указатель
    while current_weight > s:
        current_weight -= w[l]
        current_value -= c[l]
        l += 1

    # обновляем максимум стоимости
    max_value = max(max_value, current_value)

print(max_value)


# ВАШ КОД
````

## Задача H. A-B Рюкзак ([ссылка](https://codeforces.com/edu/course/2/lesson/9/3/practice/contest/307094/problem/H)) [0.5 балла]

Даны:
- 𝑛 предметов, каждый имеет вес 𝐴 и стоимость a₁, …, aₙ.
- 𝑚 предметов, каждый имеет вес 𝐵 и стоимость b₁, …, bₘ.

Из этих предметов формируют набор для рюкзака, который может выдерживать суммарный вес не более 𝑠. Требуется определить, какую максимальную суммарную стоимость можно получить, заполнив рюкзак выбранными предметами.

---

## Входные данные

- Первая строка содержит пять целых чисел: 𝑛, 𝑚, 𝑠, 𝐴 и 𝐵  
  (1 ≤ 𝑛, 𝑚 ≤ 10⁵, 1 ≤ 𝑠, 𝐴, 𝐵 ≤ 10⁹).
- Во второй строке содержатся 𝑛 целых чисел a₁, …, aₙ (1 ≤ aᵢ ≤ 10⁹) — стоимости предметов первого набора.
- В третьей строке содержатся 𝑚 целых чисел b₁, …, bₘ (1 ≤ bᵢ ≤ 10⁹) — стоимости предметов второго набора.

---

## Выходные данные

Выведите одно число — наибольшую суммарную стоимость предметов, которые можно поместить в рюкзак, так чтобы суммарный вес не превышал 𝑠.

---

## Пример

**Входные данные:**
```
6 7 23 3 5
7 4 3 1 5 8
10 12 7 3 8 9 7
```

**Выходные данные:**
```
47
```

````python
# Асимптотическая сложность по времени - O(n log n + m log m)
# Асимптотическая сложность по памяти - O(n + m)

# ВАШ КОД
n, m, s, A, B = map(int, input().split())
a = list(map(int, input().split()))
b = list(map(int, input().split()))

# Сортируем по убыванию стоимости
a.sort(reverse=True)
b.sort(reverse=True)

# Префиксные суммы для обоих массивов
prefix_a = [0] * (n + 1)
prefix_b = [0] * (m + 1)
for i in range(n):
    prefix_a[i + 1] = prefix_a[i] + a[i]
for i in range(m):
    prefix_b[i + 1] = prefix_b[i] + b[i]

# Перебираем количество предметов типа A
max_cost = 0
j = m
for i in range(0, n + 1):
    total_weight_A = i * A
    if total_weight_A > s:
        continue
    while j > 0 and total_weight_A + j * B > s:
        j -= 1
    total_cost = prefix_a[i] + prefix_b[j]
    max_cost = max(max_cost, total_cost)

print(max_cost)

````

### ВАША интуиция

Отсортируем, потом сделаем перефексную сумму, потом сделаем линейный проход двумя указателями
